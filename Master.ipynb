**Generate Figure 2(b)**

For Figure 2(b), set `use_solver=True`

To plot previously-generated outputs for Figure 2(b), use `skip_calc=True`

```python
# DEFAULT INPUTS in .py file
# !python run_calc_min_ch4.py

# CUSTOM INPUTS
import numpy as np
import importlib

import fair_utils
import run_calc_min_ch4
importlib.reload(fair_utils)
importlib.reload(run_calc_min_ch4)

def generate_ex_ante_distribution(n_projects=10000,
                      ramp_min=5,
                      ramp_max=15,
                      total_mass=1.0,
                      sim_horizon=50):
    """
    Generates a probability distribution of net CO₂ drawdown schedules based on
    random linear ramp-up periods.
    """
    ensemble_profile = np.zeros(sim_horizon)
    t_bounds = np.arange(sim_horizon + 1)

    for _ in range(n_projects):
        D = np.random.uniform(ramp_min, ramp_max)
        slope = (2 * total_mass) / (D**2)

        cum_mass_at_boundaries = np.where(
            t_bounds < D,
            0.5 * slope * t_bounds**2,
            total_mass
        )

        yearly_flux = np.diff(cum_mass_at_boundaries)
        ensemble_profile += yearly_flux

    prob_distribution = ensemble_profile / n_projects
    return prob_distribution

print("Generating Ex-Ante Profile...")
ex_ante_profile = generate_ex_ante_distribution(
    n_projects=1,
    ramp_min=10,
    ramp_max=10,
    total_mass=1.0,
    sim_horizon=50
)

# Build custom scenario arrays
profile_len = len(ex_ante_profile)
offset_idx = fair_utils.OFFSET
custom_co2_cdr = np.zeros(1000)
custom_co2_cdr[offset_idx + 10 : offset_idx + 10 + profile_len] = -ex_ante_profile
custom_co2_emission = np.zeros(1000)
custom_co2_emission[offset_idx] = 1.0

# Run the CH4 calculator
run_calc_min_ch4.execute_calc_min_ch4(
    co2_emission_adj=custom_co2_emission,
    co2_cdr_adj=custom_co2_cdr,
    GWP_horizon=20,
    use_solver=True,
    tolerance=0.01,
    skip_calc=True
)
```
